**3.Agriculture-Rainfall Adequacy Checker**

```python
def average_rainfall():

  days=int(input("Enter number of days: "))
  total_rainfall=0

  for i in range(days):
    rainfall=float(input("Enter rainfall for day" + str(i+1)+ ":"))
    total_rainfall+=rainfall

  average=total_rainfall/days
  required_rainfall=float(input("Enter required rainfall level: "))

  if average >=required_rainfall:
    status="Adequate Rainfall"
  else:
    status="Inadequate Rainfall"

  print("Average rainfall: ",average)
  print("Rainfall Status: ",status)

average_rainfall()

```

**9.E-Learning -Student Attendance Eligibility System**

```python
list=["P", "P", "A", "P", "P"]
def attendance_eligibility(list):
  present_count = list.count("P")
  total_classes=len(list)
  attendance_percentage=(present_count/total_classes)*100
  if attendance_percentage >= 75:
    status="Eligible"
  else:
    status="Not Eligible"
  return attendance_percentage, status
percentage ,status=attendance_eligibility(list)
print("Attendance Percentage: ",percentage)
print("Exam Eligibility: ", status)


```

**10.Smart Electricity Bill Analyzer**

```python
def electricity_bill(units):
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"
    return bill, status
units = int(input("Enter electricity units consumed: "))
bill, status = electricity_bill(units)
print("Total Bill: ₹", bill)
print("Usage Level:", status)
```
