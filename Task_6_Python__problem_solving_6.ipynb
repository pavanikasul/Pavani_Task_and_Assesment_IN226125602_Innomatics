**1.Smart Parking Lot Management System**

```python
def smart_parking(capacity, logs):
  current=0
  peak=0
  for log in logs:
    if log =="IN":
      current+=1
    elif log =="OUT":
      current-=1
    if current > peak:
      peak=current
    if current>capacity:
      print("Alert: Parking capacity exceeded!")

  if current<capacity:
    status="Available"
  else:
    status="Full"

  print("Total Vechicles currently parked:",current)
  print("Parking status: ",status)


capacity=50
logs=["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
smart_parking(capacity, logs)
```

**2.Online Food Delivery Time Estimator**

```python
def delivery_time(distance,traffic,weather):
  speed=30
  traffic=traffic.lower()
  if traffic=="high":
    speed*=0.5
  elif traffic=="medium":
    speed*=0.75
  elif traffic=="low":
    speed*=1

  time=distance/speed
  time_minutes=time*60

  weather=weather.lower()
  if weather=="rainy":
    time_minutes+=10
  elif weather=="fog":
    time_minutes+=15
  elif weather=="storm":
    time_minutes+=20

  return round(time_minutes)
distance=8 #km
traffic="high"
weather="Rainy"

t=delivery_time(distance, traffic, weather)
print("Estimated time: ",t,"minutes")
```

**3.Movie Theatre Seat Occupancy Analyzer**

```python
def theatre_analyzer(total_seats, booked_seats):
  occupancy_percentage=(booked_seats/total_seats)*100

  if occupancy_percentage == 100:
    status="House Full"
  elif occupancy_percentage >=75:
    status="Almost Full"
  elif occupancy_percentage>=50:
    status="Still 100 seats are left"
  else:
    status="seats are available"
  return occupancy_percentage,status

total_seats=200
booked_seats=150
occupancy, status=theatre_analyzer(total_seats, booked_seats)
print("Occupancy: ",round(occupancy,2),"%")
print("Show Status: ",status)



```

**4.Cloud Server Load Classification System**

```python
def server_load(cpu_readings):
  total=0
  for readings in cpu_readings:
    total+=readings
    if readings <0.50:
      status="Normal"
    elif 0.50<=readings<80:
      status="Warning"
    else:
      status="Critical"



  average=total/len(cpu_readings)
  print("Server status: ",status)
  print("Average CPU Load: ",round(average,2))

cpu_readings=[0.45, 0.60, 0.70, 0.85, 0.90]
server_load(cpu_readings)
```

**5.Smart Classroom Resource Usage Monitor**

```python
def classroom_resources(resource_usage):
  limit=8
  overused=[]
  for resource, hours in resource_usage.items():
    if hours >limit:
      overused.append(resource)
  if overused:
    alert="Yes"
  else:
    alert="No"

  if overused:
    print("Overused Resources: ",",".join(overused))
  else:
    print("Overused Resources: None")

  print("Energy Alert: ",alert)
usage={
 "Projector": 6,
 "AC": 9,
 "Lights": 4
}
classroom_resources(usage)
```

**6.Online Event Registration Capacity Controller**

```python
def event_registration(registration, capacity):
  confirmed=0
  waitlist=0
  for i in range(registration):
    if confirmed < capacity:
      confirmed+=1
    else:
      waitlist+=1
  if confirmed==capacity:
    status="closed"
  else:
    status="open"

  print("Confirmed registrations: ",confirmed)
  print("Waitlist Users: ",waitlist)
  print("Registration Status: ",status)

capacity= 100
registration= 105

event_registration(registration, capacity)
```
